**SECTION 2 EXPLORATORY DATA ANALYSIS (EDA)**

---



```python
# Transaction Year Analysis
transactions['transaction_year'] = transactions['transaction_date'].dt.year
transactions['transaction_year'].value_counts().sort_index()
```

```python
# Transaction Amount By Year
yearly_summary= transactions.groupby('transaction_year')['amount_usd'].agg(
    total_transaction_value='sum',
    average_transaction_value='mean'
)
yearly_summary.style.format({
    'total_transaction_value': '${:,.2f}',
    'average_transaction_value': '${:,.2f}'
}).highlight_max(
subset=['total_transaction_value','average_transaction_value'],
props='background-color: white; color: darkgreen; font-weight: bold;'
)
```

```python
# Monthly Transaction Analysis
transactions['transaction_month'] = transactions['transaction_date'].dt.month
monthly_summary = transactions.groupby('transaction_month')['amount_usd'].agg(
    total_transaction_value='sum',
    average_transaction_value='mean'
)
monthly_summary.style.format({
    'total_transaction_value': '${:,.2f}',
    'average_transaction_value': '${:,.2f}'
}).highlight_max(
subset=['total_transaction_value','average_transaction_value'],
props='background-color: white; color: darkgreen; font-weight: bold;'
)
```

```python
# Transaction Analysis BY Account Type
transactions_account = transactions.merge(
    accounts[['account_id','account_type']],
    on = 'account_id',
    how = 'left'
    )

transactions_account.head().style.format({
    'amount_usd': '${:,.2f}'
}).highlight_max(
subset=['amount_usd'],
props='background-color: white; color: darkgreen; font-weight: bold;'
)
```

```python
# Analyze Transactions By Account Type
account_type_summary = transactions_account.groupby('account_type')['amount_usd'].agg(
    total_transactions = 'count',
    total_transaction_value='sum',
    average_transaction_value='mean'
)
account_type_summary.style.format({
    'total_transactions': '{:,.0f}',
    'total_transaction_value': '${:,.2f}',
    'average_transaction_value': '${:,.2f}'
}).highlight_max(
subset=['total_transaction_value','average_transaction_value'],
props='background-color: white; color: darkgreen; font-weight: bold;'
)
```

```python
# Customer Credit Score + Loan Analysis
customer_credit_loan = customers[
    ['customer_id','credit_score']
].merge(
    customer_loan_analysis[
        ['customer_id','total_loans','total_loan_amount','average_interest_rate']
    ],
    on = 'customer_id',
    how = 'left'
)
customer_credit_loan[
    ['total_loans','total_loan_amount']
] = customer_credit_loan[
    ['total_loans','total_loan_amount']
].fillna(0)
customer_credit_loan['credit_category'] = pd.cut(
    customer_credit_loan['credit_score'],
    bins = [299,579,669,739,799,850],
    labels = ['Poor','Fair','Good','Very Good','Excellent']
)
credit_loan_analysis = customer_credit_loan.groupby(
    'credit_category',observed=False
).agg(
    customers = ('customer_id','nunique'),
    average_credit_score = ('credit_score','mean'),
    average_loan_amount = ('total_loan_amount','mean'),
    average_interest_rate = ('average_interest_rate','mean')
).round(2)
credit_loan_analysis.style.format({
    'customers': '{:,.0f}',
    'average_credit_score': '{:.2f}',
    'average_loan_amount': '${:,.2f}',
    'average_interest_rate': '{:.2f}%'
}).highlight_max(
    subset=['average_loan_amount'],
    props='background-color: white; color: darkgreen; font-weight: bold;'
)

```
